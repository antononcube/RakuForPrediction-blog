## Introduction

This document ([notebook]()) discusses number theory properties and relationships of the integer 2027.

The integer `2027` is prime -- also it is a safe-, strong-, twin-, and prime-indexed prime, with 2 as one of its primitive roots. So, `2027` is a fairly notable prime - the combination of being a twin prime, a safe prime, and possessing a small primitive root (namely 2) is comparatively rare among four-digit primes. Certain elaborate visualizations can be made that reveal some interesting aspects of the number.

The computations in this document are done with the Raku package ["Math::NumberTheory"](https://raku.land/zef:antononcube/Math::NumberTheory), [AAp1].

## 2027 Is a Notable Prime

First, 2027 is obviously prime — it can be represented as `2 * 1013 + 1` (and `1013` is prime, as discussed [here](https://raku-advent.blog/2025/12/22/day-22-numerically-2026-is-unremarkable-yet-happy/), [AA3]):

```raku
is-prime(2027)
```

`2027` is also a [Gaussian prime](https://en.wikipedia.org/wiki/Gaussian_integer#Gaussian_primes) (but the integer `1013` is not):

```raku
say is-prime(2027, :gaussian-integers);
say is-prime(1013, :gaussian-integers);
```

A [**safe prime**](https://en.wikipedia.org/wiki/Safe_and_Sophie_Germain_primes) $p$ is a number which can be represented as $p = 2 q + 1$, where $q$ is also a prime -- those $q$'s are called **Sophie Germain primes**.  

As shown above `2027` is a safe prime.

Moreover, `2027` is a prime-indexed prime -- $2027$ is the $307$-th prime, and the index $307$ is also prime:

```raku
say prime(307);
say is-prime(307);
```

Not many 4-digit prime numbers have all these properties:

```raku
my @notable-primes = 
(168 ..^ 1230).grep({
    my $p = prime($_);
    is-prime($p) &&           # Prime number
    is-prime($p + 2) &&       # Twin prime
    is-prime(($p - 1) / 2) && # Safe prime
    (prime($_ - 1) + prime($_ + 1)) / 2 < $p # Strong prime
})».&prime();

say (:@notable-primes);
say "4-digit primes count: {1230 - 168}, notable primes count: {@notable-primes.elems}, fraction: {@notable-primes.elems / (1230 - 168)}"
```

Being prime means the multiplicative group modulo `2027` has primitive roots. A primitive root exists for an integer $n$ if and only if $n$ is $1$, $2$, $4$, $p^k$, or $2 p^k$, where $p$ is an odd prime and $k > 0$.

Let us compare $2027$ properties with those of the [feature-rich 2025](https://rakuforprediction.wordpress.com/2025/01/03/numeric-properties-of-2025/) 
and the just [happy 2026](https://raku-advent.blog/2025/12/22/day-22-numerically-2026-is-unremarkable-yet-happy/).
(In the next section.)

## Comparison with 2025 and 2026

Here is a side-by-side comparison of key number theory properties for `2025`, `2026`, and `2027`.

| Property                  | 2025                                      | 2026                                      | 2027       |
|---------------------------|-------------------------------------------|-------------------------------------------|------------|
| **Prime or Composite**    | Composite                                | Composite                                | Prime |
| **Prime Factorization**   | 3⁴ × 5² (81 × 25)                        | 2 × 1013                                 | 2 * 1013 + 2 (safe prime) |
| **Number of Divisors**    | 15 (highly composite for its size)       | 4 (1, 2, 1013, 2026)                     | 2 (1, 2027) |
| **Perfect Square**        | Yes (45² = 2025)                         | No                                       | No |
| **Sum of Cubes**          | Yes (1³ + 2³ + ... + 9³ = (1 + ... + 9)² = 2025)          | No                      | No |
| **Happy Number**          | No (process leads to cycle including 4)  | Yes (repeated squared digit sums reach 1) | No |
| **Harshad Number**        | Yes (divisible by 9)                      | No (not divisible by 10)                 | No (not divisible by 11) |
| **Primitive Roots**       | No                                       | Yes                                      | Yes |
| **Other Notable Traits**  | - (20 + 25)² = 2025<br>- Sum of first 45 odd numbers<br>- Deficient number<br>- Many pattern-based representations | - Even number<br>- Deficient number<br>- Few special patterns | prime indexed, safe, strong, twin prime |
| **Overall "Interest" Level** | Highly interesting—celebrated in math communities for squares, cubes, and patterns | Relatively uninteresting—basic semiprime with no standout geometric or sum properties | 👆Reinforces blog's angle. |

To summarize:

- **2025** stands out as a mathematically rich number, often highlighted in puzzles and articles for its perfect square status and connections to sums of cubes and odd numbers.  
- **2026** has fewer flashy properties -- it's a straightforward even semiprime -- but it qualifies as a *happy number* and it has a primitive root.
- **2027**, in contrast, is prime, with lots of properties that make it a *notable prime*. 

Here is a computationally generated comparison table of most of the properties found in the table above:

```raku
#% html
my &divisors-count = { divisors($_).elems };
my &is-twin-prime = { is-prime($_) && is-prime($_ + 2) };
<is-prime is-composite divisors-count prime-omega euler-phi is-square-free is-happy-number is-harshad-number is-deficient-number primitive-root is-twin-prime>.map({ %(sub => $_, '2025' => ::("&$_")(2025), '2026' => ::("&$_")(2026), '2027' => ::("&$_")(2027) ) })
==> to-html(field-names => ['sub', '2025', '2026', '2027'], align => 'left')
```

```raku
from-number-expansion(@res, 'Zeckendorf')
```

```raku
fibonacci((2...(@res.elems+1))) <<*>> @res.reverse
==> sum()
```

## Triangular Numbers

There is a theorem by Gauss stating that any integer can be represented as a sum of at most three triangular numbers. Instead of programming such an integer decomposition representation in Raku, we can simply use [Wolfram|Alpha](https://www.wolframalpha.com), [AA1, AA3], or [`wolframscript`](https://www.wolfram.com/wolframscript/) to find an "interesting" solution:

```raku
#% bash
wolframscript -code 'FindInstance[{2027 == PolygonalNumber[i] + PolygonalNumber[j] + PolygonalNumber[k], i > 10, j > 10, k > 10}, {i, j, k}, Integers]'
```

Here, we verify the result using Raku:

```raku
say "Triangular numbers : ", <13 43 44>.&polygonal-number(:3sides);
say "Sum : ", <13 43 44>.&polygonal-number(:3sides).sum;
```

## Perfect square graph

```raku
my @ns = 1900...2027;
my @edges = cross(@ns, @ns).map({ my $s = $_.head * $_.tail + 1; $s.sqrt.Int == $s.sqrt ?? ($_.head.Str => $_.tail.Str) !! Nil }).grep(* ~~ Pair);
deduce-type(@edges)
```

```raku
my @ns = 1900...2027;
my @edges = @ns.map({ $_.Str => divisors($_).sum.Str }).grep(* ~~ Pair);
deduce-type(@edges)
```

```raku
#% html
my @prime-too = @ns.grep({ is-prime($_) && $_ != 2027 });

my %highlight = '#006400' => @prime-too».Str,      # Deep Christmas green for primes
                'Blue' => [2027.Str, ];              # Blue for the special year

my $gRoots = Graph.new(@edges):directed;

$gRoots.dot(
    engine => 'neato', 
    graph-size => 12, 
    vertex-shape => 'ellipse', vertex-height => 0.08, vertex-width => 0.17,
    :5vertex-font-size,
    vertex-color => '#B41E3A',
    vertex-fill-color => '#B41E3A',
    arrowsize => 0.2,
    edge-color => 'gray',
    edge-width => 0.5,
    splines => 'line',
    :%highlight
):svg
```

## Primitive Root Connections Graph

Let us create and plot a graph showing the trails of all happy numbers less than or equal to 2026. Below, we identify these numbers and their corresponding trails:

```raku
my @ns = 1900...2027;
my %primitive-roots = @ns.map({ $_ => primitive-root-list($_).Set });
deduce-type(%primitive-roots)
```

```raku
my @edges = cross(@ns, @ns).map({ $_.head ∈ %primitive-roots{$_.tail} ?? ($_.head.Str => $_.tail.Str) !! Nil }).grep(* ~~ Pair);
deduce-type(@edges)
```

Here is the corresponding connections graph, highlighting the primes and numbers with primitive roots:

```raku
#% html
my @prime-too = @ns.grep({ is-prime($_) && $_ != 2027 });
my @with-roots = @ns.grep({ %primitive-roots{$_} && !is-prime($_) });

my %highlight = '#006400' => @prime-too».Str,      # Deep Christmas green for primes
                'Blue' => [2027.Str, ],              # Blue for the special year
                '#fbb606ff' => @with-roots».Str;   # Darker gold for composite integers with primitive roots

my $gRoots = Graph.new(@edges):directed;

$gRoots.dot(
    engine => 'twopi', 
    graph-size => 12, 
    vertex-shape => 'ellipse', vertex-height => 0.08, vertex-width => 0.17,
    :5vertex-font-size,
    vertex-color => '#B41E3A',
    vertex-fill-color => '#B41E3A',
    arrowsize => 0.2,
    edge-color => 'gray',
    edge-width => 0.1,
    splines => 'line',
    :%highlight
):svg
```

```raku
#`[
#%  js
$gRoots.edges ==> 
js-d3-graph-plot(
    :$background,
    :%highlight,
    :800width,
    :800height,
    vertex-size => 4,
    vertex-color => 'Red',
    force => {charge => {strength => -30, iterations => 4}, collision => {radius => 35, iterations => 4}, link => {distance => 3}}
)
]
```

## Known sequence membership

The number `2027` appears in [702 results of the search "2027"](https://oeis.org/search?q=2027&fmt=data) in [«The On-line Encyclopedia of Integer Sequences»](https://oeis.org/). Here is the first result (from 2026-11-25): [A235395](https://oeis.org/A235395), "
Primes whose decimal representation is a valid number in base 9 and interpreted as such is again a prime." Below, we display the prime numbers in A235395 in red, the prime numbers in green, anything else in white using Ulam's spiral:

```raku
#%html
my $n = 13;
my @base9-primes = (1 .. $n²).grep({ is-prime($_) && is-prime($_.base(9).Int) });
my @other-primes = (1 .. $n²).grep({ is-prime($_) && !is-prime($_.base(9).Int) });
spiral-lattice($n, :dataset)
==> to-html(field-names => (^$n)».Str)
==> html-table-highlight(highlight => @base9-primes, color => 'Red', font-weight => 'bold')
==> html-table-highlight(highlight => @other-primes, color => 'Green', font-weight => 'bold')
==> html-table-highlight(highlight => (1 .. $n²), color => 'DarkGray')
```

Here is a visualization of the same groups of numbers for larger span of integers -- 2027 is marked with the brightest square:

```raku
#% js
my @mat = spiral-lattice(101).deepmap({ 
    given $_ {
        when 2027 { 12 }
        when is-prime($_) && is-prime($_.base(9).Int) { 8 }
        when is-prime($_) { 4 }
        default {0}
    }
 });
js-d3-matrix-plot(@mat, width => 500, height => 500, :!grid-lines, :!tooltip, color-palette => 'Magma')
```
