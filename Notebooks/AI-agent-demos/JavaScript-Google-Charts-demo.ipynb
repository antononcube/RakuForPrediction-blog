## Population and land area

This Bubble chart directly compares each city's land area and population. Bubble size encodes population density (people per square mile).

```raku
#% html
my @area-population = @usa-cities.map: -> %city {
    {
        city => %city<city>,
        area => %city<area>,
        population => %city<population>,
        group => 'U.S. city',
        density => %city<population> / %city<area>
    }
};

js-google-charts('Bubble', @area-population,
    column-names => <city area population group density>,
    format => 'html',
    title => 'Population versus land area of the ten most populous U.S. cities',
    backgroundColor => '#111827',
    chartArea => { backgroundColor => '#111827' },
    titleTextStyle => { color => '#f8fafc' },
    hAxis => { title => 'Land area (square miles)', titleTextStyle => { color => '#e2e8f0' }, textStyle => { color => '#cbd5e1' }, gridlines => { color => '#334155' }, baselineColor => '#64748b' },
    vAxis => { title => 'Population (2020 Census)', titleTextStyle => { color => '#e2e8f0' }, textStyle => { color => '#cbd5e1' }, gridlines => { color => '#334155' }, baselineColor => '#64748b' },
    legend => 'none',
    colors => ['#38bdf8'],
    bubble => { textStyle => { color => '#f8fafc', fontSize => 11 } },
    sizeAxis => { minSize => 12, maxSize => 55 },
    width => 900, height => 600,
    div-id => 'usa-cities-area-population');
```

## Number theory

The sunflower (phyllotaxis) layout places integer $n$ at radius $\sqrt{n}$ and angle $n$ times the golden angle. Colour and bubble size both encode the residue modulo 5; sizes use residue + 1 so multiples of 5 remain visible.

```raku
#% html
my $golden-angle = pi * (3 - sqrt(5));
my @sunflower = (1 .. 5000).map: -> $n {
    my $remainder = $n % 5;
    my $radius = sqrt($n);
    my $angle = $n * $golden-angle;
    {
        integer => $n.Str,
        x => $radius * cos($angle),
        y => $radius * sin($angle),
        remainder => $remainder,
        size => $remainder + 1
    }
};

js-google-charts('Bubble', @sunflower,
    column-names => <integer x y remainder size>,
    format => 'html',
    title => 'Sunflower distribution of integers 1-5000 (mod 5)',
    backgroundColor => '#121212',
    chartArea => { backgroundColor => '#121212', left => 10, top => 42, width => 880, height => 928 },
    titleTextStyle => { color => '#77787a' },
    hAxis => { textPosition => 'none', gridlines => { color => 'transparent' }, baselineColor => 'transparent' },
    vAxis => { textPosition => 'none', gridlines => { color => 'transparent' }, baselineColor => 'transparent' },
    legend => 'none',
    colorAxis => { colors => ['#38bdf8', '#a78bfa', '#f472b6', '#fb923c', '#facc15'], minValue => 0, maxValue => 4, legend => { position => 'none' } },
    bubble => { opacity => 0.75 },
    sizeAxis => { minSize => 2, maxSize => 9 },
    width => 700, height => 700,
    div-id => 'mod-5-sunflower');
```
