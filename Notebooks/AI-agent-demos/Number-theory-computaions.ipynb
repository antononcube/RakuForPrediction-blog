## First go

```raku
#%js
my @carmichael-lambda-matrix = (1..100).map(-> $j {
    (1..100).map(-> $k { carmichael-lambda($j + 5 * $k) }).Array
}).Array;

#% js
js-d3-matrix-plot(
    @carmichael-lambda-matrix,
    width => 600, height => 600,
    :!grid-lines, :!tooltip, :!mesh,
    color-palette => 'Turbo'
)
```

## Second go

```raku
#% js
# Plot the Mangoldt lambda function for sums of two squares.
my @mangoldt-lambda-matrix = (1..50).map(-> $a {
    (1..50).map(-> $b { mangold-lambda($a ** 2 + $b ** 2) }).Array
}).Array;

#%js
js-d3-matrix-plot(
    @mangoldt-lambda-matrix,
    width => 400, height => 400,
    :!grid-lines, :!tooltip, :!mesh,
    color-palette => 'YlOrBr'
)
```

```raku
#%js

# Plot Mangoldt lambda values on a 101 x 101 Ulam spiral.
my @ulam-mangoldt-matrix = spiral-lattice(101).map({
    $_.map(&mangold-lambda).Array
}).Array;

#%js
js-d3-matrix-plot(
    @ulam-mangoldt-matrix,
    width => 450, height => 450,
    :!grid-lines, :!tooltip, :!mesh,
    color-palette => "Turbo"
)
```
